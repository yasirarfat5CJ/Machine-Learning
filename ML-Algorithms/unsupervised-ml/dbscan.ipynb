```python
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt

# Load Iris dataset
iris = load_iris()
X = iris.data

# Scale features (VERY important for DBSCAN)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# DBSCAN model
db = DBSCAN(eps=0.6, min_samples=5)
labels = db.fit_predict(X_scaled)

print("Cluster labels:", labels[:20])  # -1 = noise

# Plot results (using first 2 features)
plt.scatter(X_scaled[:,0], X_scaled[:,1], c=labels, cmap='rainbow')
plt.title("DBSCAN Clustering (Iris Dataset)")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```
